## Data Mining x Machine Learning

- Data Mining:
    - Descoberta de conhecimento a partir dos dados;
    - Descobrir padrões e modelos que sejam:
        - Valid
        - Useful
        - Unexpected
        - Undestandable
    - Parte do processo de KDD (Knowledge Discovery in Databases (KDD))
> Ex.: Dado os atributos de um determinado contribuinte, classifica-lo como possível sonegador ou não.
- Machine Learning:
    - Processo de descoberta de algoritmos que melhoraram graças à experiência derivada dos dados ou de suas execuções anteriores.
    - É o projeto, estudo e desenvolvimento de algoritmos que permitem que as máquinas aprendam sem intervenção humana. 
    - "É o campo de estudo que ter por finalidade aos computadores a habilidade de aprender sem serem explicitamente programados" (Arthur Samuel, 1959)
> Um algoritmo que aprende sozinho (ou vai melhorando com o tempo) a classificar um contribuinte como possível sonegador ou não.


## Data Science

É a ciência que usa ciência, estatística e ML, visualização e interação humano-computador para coletar, integrar, analisar, visualizar, interagir com a data para recriar produtos de data. "Turn data into data products"

## O que é Data Science?

A Ciência de Dados é uma área multidisciplinar que orienta a **extração de informação** e **conhecimento** a partir de **grandes volumes de dados** [Provost and Fawcett 2013].

Trata da **coleta**, **integração**, **gerenciamento**, **exploração** dos conjuntos de dados e da utilização do conhecimento adquirido com a finalidade de **tomar decisões**, **entender o passado/presente**, **prever o futuro** e **criar novos serviços/produtos** [Ozdemir 2016].

A Ciência de Dados busca obter novas ideias (**“insights”**) que estejam escondidas nesses grandes repositórios de dados.


## Taxonomia
### Métodos de ML/DM

1. Aprendizado Supervisionado:
    A característica básica de sistemas de aprendizado supervisioado é que os dados que utilizamos para treiná-los contém a resposta desejada, ou seja, variável dependete (target value ou class label).

2. Aprendizado Não-Supervisionado:
    Em aprendizado não supervisionado os dados não possuem rótulo. Assim o algoritmo de ML irá buscar padrões existentes no conjunto de dados (Cluster).

        - Aprendizado não-supervisionado também é importante para detecção de anomalias(outliers), como por exemplo: detecção de transação de cartão de crédito. 

## Aprendizado Semi-Supervisionado

- Rotular os dados implica na participação de especialistas(de forma não-automática)
- Em problemas reais, pode ser bastante caro ou mesmo inviável rotular todos os dados coletados:
    - Rotular 270.000 mensagens capturadas do WhatsApp como desinformação ou não;
- Nestes casos, temos:
    - Uma pequena parte dos dados está rotulada
    - A maior parde dos dados não está rotulada
- Porém:
    - Os exemplos não rotulados podem ser úteis para melhorar a performance dos algoritmos supervisionados;
    - Os exemplos rotulados podem ser utilizados para rotular os demais de forma automática;

## Aprendizado por Reforço

- Consiste no treinameendo de modelos para tomar uma sequência de decisões;
- **O modelo:**
  - Aprende a atingir uma meta em um ambiente incerto e potencialmente complexo;
  - Enfrenta um problema (tem que tomar uma decisão/ação);
  - Inicialmente, utiliza tentativa e erro para encontrar uma solução para o problema;
  - Recebe recompensas ou penalidades pelas ações que executa.  
    Seu objetivo é maximizar a recompensa total.



- **Formalmente, o modelo consiste de:**
  - Um conjunto discreto de estados do ambiente, **S**;
  - Um conjunto discreto de ações do agente, **A**;
  - Um conjunto de sinais de reforço, normalmente **{0,1}**, ou um número real.

- **Exemplos:**
  - Jogar um determinado jogo;
  - Equilibrar um robô ou drone;


## Métodos preditivos x Métodos descritivos
- **Métodos Preditivos**
  - Usa os valores de algumas variáveis para prever valores desconhecidos ou futuros de outras variáveis.

- **Métodos Descritivos**
  - Encontrar padrões interpretáveis por humanos que descrevem os dados.


## Regressão
Prever número.  
Ex.: Prever o valor de um carro/casa.
- Principais Métodos/Algoritmos (Preditores):
    1. Regressão Linear Simples
    2. Regressão Linear Múltipla
    3. Regressão Polinomial

## Classificação
Classificar um registro quanto ao atributo classe a partir dos dados valores dos outros atributos
Ex.: Realizar uma pesquisa para ver quais clientes estavam interessados no novo modelo de um carro

- Principais Métodos/Algoritmos (Classificadores):

    - Logistic Regression
    - Decision Trees
    - Random Forest
    - kNN
    - Redes Neurais
    - Support Vector Machines (SVMs)



### Clusterização
Busca encontrar grupos (clusters) de instâncias (observações ou registros), tais que:

- Duas instâncias quaisquer que pertencem a um mesmo grupo (cluster) possuem alta similaridade;

    - Muito semelhantes entre si;

- Duas instâncias quaisquer que pertencem a grupos (clusters) distintos possuem baixa similaridade;

    - Muito distintas entre si;

Ex.: Encontrar grupos de consumidores; agrupar documentos; compressão de dados

#### Principais métodos/algoritmos
- k-Means
- Hieraquical Cluster Analysis (HCA)
- Expectation Maximization
- CURE Algorithm

### Regras de Associação
- Busca encontrar padrões nos dados que permitam identificar atributos (features) que estejam fortemente associados;

    - Tais padrões são tipicamente representados por meio de regras de implicação ou subconjuntos de features;

    - Como o espaço de busca é exponencial (em relação ao número de features);

        - Busca-se extrair os padrões mais interessantes de uma maneira eficiente;

#### Principais métodos/algortimos:
- Apriori
- Eclat


### Detecção de Anomalias
- Principais Métodos/Algoritmos:
    - kNN

    - SVM

    - Redes Neurais

    - Redes Bayesianas

    - Modelos ocultos de Markov (HMMs)



### Redução de Dimensionalidade
Os datasets podem ter um grande número de features (atributos);

> Ex: Um conjunto de documentos, onde cada coluna representa a quantidade de vezes que uma determinada palavra ocorre no documento;

> Muitos algoritmos de ML trabalham melhor se o número de features for pequeno;

- Benefícios:
    - Pode eliminar features irrelevantes;

    - Pode reduzir ruído;

    - Pode levar a modelos mais fáceis de compreender;

    - Os dados podem ser visualizados de maneira mais simples e efetiva;

    - Reduz o tempo de processamento para geração de modelos e a quantidade de memória utilizada;

- Principais Estratégias:
    - Seleção de um subconjunto dos atributos
        - Remover features redundantes ou irrelevantes;
        - Principais abordagens;
            - Filter;
            - Wrapper;
            - Embedded;
        - Criação de um conjunto de novos atributos gerados a partir da combinação dos atributos originais;

- Principais Métodos/Algoritmos:
    - Principal Component Analysis (PCA)

    - Kernel PCA

    - Locally-Linear Embedding (LLE)

    - t-distributed Stochastic Neighbor Embedding (t-SNE)

    - Single Value Decomposition (SVD)
