- O que é estatística?
É a ciência que trata da coleta, organização, análise dos dados para tomada de decisões
coletar dados -> analisar dados -> interpretar os dados -> apresentar resultados.
- estatística e IA:
A maioria dos algoritmos de ML é, essencialmente, estatística computacional aplicada.
Ex.: Regressão linear e logística.

Ex.: validação cruzada, intervalo de confiança para acurácia e testes de hipóteses entre modelos.

### Conceitos básicos

- Dados: são dispostos normalmente na forma de uma tabela nxp, com n indivíduos (ou amostras) e p variáveis. Ex.: n>>p | n << p | n e p grandes.

- Tipos de variáveis:
**Qualitativa**: Possuem valores que são qualidades, ex.: Tipo de contrato, tamanho de camisa, grau de satisfação.
> As variáveis qualitativas **nominais não possuem ordem** inerente.

> As variáveis qualitativas **ordinais poosuem uma ordem implícita**

**Quantitativa**: possuem como valores números resultantes de uma contagem ou mensuração. Ex.: idade, altura, peso, etc.

> Discretas possuem valores que formam um conjunto finito ou enumerável de números. Normalmente, resultam de uma contagem

> Contínuas possuem valkres possíveis pertencentes a um intervalo de números reais. Normalmente resultam de uma medição.


- Estado civil: Qualitativa Nominal
- Grau de instrução: Qualitativa Ordinal
- N° de filhos: Quantitativa Discreta
- Salário: Quantitativa Contínua
- Idade: Quantitativa Discreta
- Região de Procedência: Qualitativa Nominal

1. Variáveis dicotômicas (booleanas) só possuem dois valores possíveis, sucesso ou fracasso / verdadeiro ou falso, etc.
2. Dados de seção transversal e de série temporal 
> Transversal são aqueles coletados no mesmo ou aproximadamente no mesmo intervalo de tempo

> Série temporal são aqueles coletados ao longo de diversos períodos.

- Escalas
1. Medição: podem ser classificadas com a escala usada para realizar a medição.

    1.1. Escala Nominal: é diferente ou não de outra, categorizar os individuos.

    1.2. Escala Ordinal: Situação anterior, mas as categorias são ordenadas, e a ordem dos numerais associados ordena as categorias. Ex.: classe socioeconômica, baixa(1), média(2) e alta (3).

    1.3. Escala Intervalar: Medida é igual ou diferente, maior e quão maior do que outra. Necessitamos de uma origem arbitrária e de uma unidade de mediada. Ex.: Temperatura de um indivíduo.

    1.4. Escala Razão: podemos dizer se são iguais, ou se é difernte, maior, quão maior e quantas vezes a outra.

- Elementos: são as entidades a respeito das quais se coletam dados.

## Ramos da Estatística

**Estatística descritiva** é o ramo da estatística que envolve a organização, o resumo e a representação dos dados. A estatística descritiva existe para simplificar, o que sempre implica alguma perda de nuance ou detalhe.

**Estatística inferencial** é o ramo da estatística que envolve o uso de uma amostra para chegar a conclusões sobre uma população. Em suma, podemos usar dados do “mundo conhecido” para fazer inferências informadas sobre o “mundo desconhecido”.


## Exemplo hipotético – Estatística descritiva e inferencial

📊 **Estatística descritiva:**  
Corresponde aos **dados apresentados no gráfico**, ou seja, o fato de que **70% dos solteiros** e **90% dos casados** estavam vivos aos 65 anos.  
> É a parte que **resume e descreve os dados observados** na amostra estudada.

📈 **Estatística inferencial:**  
Entra em ação **quando usamos esses resultados para tirar conclusões sobre toda a população**, e não apenas sobre a amostra estudada.  
> Por exemplo: **inferir que, em geral, homens casados tendem a viver mais que homens solteiros**.

**Resumo:**
- **Estatística descritiva:** apresentar e resumir os dados obtidos.  
- **Estatística inferencial:** generalizar ou tirar conclusões a partir desses dados.
